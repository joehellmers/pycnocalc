```python
import pandas as pd
import numpy as np
```

```python
dfsaopaulo = pd.read_csv("../researchdata/pycnorates_cc_sao_paulo.csv",index_col=False)
dfm3y = pd.read_csv("../researchdata/pycnorates_cc_m3y.csv",index_col=False)
```

```python
dfsaopaulo.plot(x='density', y='pycno_rate', logx='true', logy='true', title="Pycno Rxn Rate vs Density")
dfm3y.plot(x='density', y='pycno_rate', logx='true', logy='true', title="Pycno Rxn Rate vs Density")
```

```python
dfsaopaulo['Log(density)'] = dfsaopaulo['density']
dfsaopaulo['Log(Rxn Rate])'] = 0
```

```python
dfsaopaulo['Log(density)']= np.log(dfsaopaulo.density)
dfsaopaulo['Log(Rxn Rate]']= np.log(dfsaopaulo.pycno_rate)
```

```python
dfplt = dfsaopaulo.plot(x='Log(density)', y='Log(Rxn Rate]',title="Log-Log Pycno Rxn Rate (Sao Paolo) vs Density")
dfplt.set_xlabel('Log(Density(gm/cm^3)')
dfplt.set_ylabel('Log(Rxn Rate(cnt/sec*cm^3)')
```
