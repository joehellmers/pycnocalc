```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
imagesDir = '/Users/hellmersjl/bitbucket/hellmers_ms/thesis_doc/Figures'
```

```python
dfsaopaulo = pd.read_csv("../researchdata/pycno_rates_C-C_WD_SaoPaulo.csv",index_col=False,skipinitialspace=True)
dfm3y = pd.read_csv("../researchdata/pycno_rates_C-C_WD_M3Y.csv",index_col=False,skipinitialspace=True)
dfrmf = pd.read_csv("../researchdata/pycno_rates_C-C_WD_RMF.csv",index_col=False,skipinitialspace=True)
dfsaopaulo['nninteraction'] = dfsaopaulo['nninteraction'].str.strip()
dfm3y['nninteraction'] = dfm3y['nninteraction'].str.strip()
dfrmf['nninteraction'] = dfrmf['nninteraction'].str.strip()
```

```python
dfsaopaulo_spvh = dfsaopaulo.query('nninteraction == "SAOPAULO-SPVH"')
dfsaopaulo_bcc_static = dfsaopaulo.query('nninteraction == "SAOPAULO-bcc-static"')
dfsaopaulo_bcc_ws = dfsaopaulo.query('nninteraction == "SAOPAULO-bcc-ws"')
dfsaopaulo_bcc_relax = dfsaopaulo.query('nninteraction == "SAOPAULO-bcc-relax"')
dfsaopaulo_fcc_static = dfsaopaulo.query('nninteraction == "SAOPAULO-fcc-static"')
dfsaopaulo_fcc_ws = dfsaopaulo.query('nninteraction == "SAOPAULO-fcc-ws"')
dfsaopaulo_fcc_relax = dfsaopaulo.query('nninteraction == "SAOPAULO-fcc-relax"')

dfm3y_spvh = dfm3y.query('nninteraction == "M3Y-SPVH"')
dfm3y_bcc_static = dfm3y.query('nninteraction == "M3Y-bcc-static"')
dfm3y_bcc_ws = dfm3y.query('nninteraction == "M3Y-bcc-ws"')
dfm3y_bcc_relax = dfm3y.query('nninteraction == "M3Y-bcc-relax"')
dfm3y_fcc_static = dfm3y.query('nninteraction == "M3Y-fcc-static"')
dfm3y_fcc_ws = dfm3y.query('nninteraction == "M3Y-fcc-ws"')
dfm3y_fcc_relax = dfm3y.query('nninteraction == "M3Y-fcc-relax"')

dfrmf_spvh = dfrmf.query('nninteraction == "RMF-SPVH"')
dfrmf_bcc_static = dfrmf.query('nninteraction == "RMF-bcc-static"')
dfrmf_bcc_ws = dfrmf.query('nninteraction == "RMF-bcc-ws"')
dfrmf_bcc_relax = dfrmf.query('nninteraction == "RMF-bcc-relax"')
dfrmf_fcc_static = dfrmf.query('nninteraction == "RMF-fcc-static"')
dfrmf_fcc_ws = dfrmf.query('nninteraction == "RMF-fcc-ws"')
dfrmf_fcc_relax = dfrmf.query('nninteraction == "RMF-fcc-relax"')


dfsaopaulo_spvh       = dfsaopaulo_spvh.rename(columns      ={"pycno_rate": "Sao Paulo-SPVH"})
dfsaopaulo_bcc_static = dfsaopaulo_bcc_static.rename(columns={"pycno_rate": "Sao Paulo-bcc-static"})
dfsaopaulo_bcc_ws     = dfsaopaulo_bcc_ws.rename(columns    ={"pycno_rate": "Sao Paulo-bcc-ws"})
dfsaopaulo_bcc_relax  = dfsaopaulo_bcc_relax.rename(columns ={"pycno_rate": "Sao Paulo-bcc-relaxed"})
dfsaopaulo_fcc_static = dfsaopaulo_fcc_static.rename(columns={"pycno_rate": "Sao Paulo-fcc-static"})
dfsaopaulo_fcc_ws     = dfsaopaulo_fcc_ws.rename(columns    ={"pycno_rate": "Sao Paulo-fcc-ws"})
dfsaopaulo_fcc_relax  = dfsaopaulo_fcc_relax.rename(columns ={"pycno_rate": "Sao Paulo-fcc-relaxed"})

dfm3y_spvh        = dfm3y_spvh.rename(columns      ={"pycno_rate": "M3Y-SPVH"})
dfm3y_bcc_static  = dfm3y_bcc_static.rename(columns={"pycno_rate": "M3Y-bcc-static"})
dfm3y_bcc_ws      = dfm3y_bcc_ws.rename(columns    ={"pycno_rate": "M3Y-bcc-ws"})
dfm3y_bcc_relax   = dfm3y_bcc_relax.rename(columns ={"pycno_rate": "M3Y-bcc-relaxed"})
dfm3y_fcc_static  = dfm3y_fcc_static.rename(columns={"pycno_rate": "M3Y-fcc-static"})
dfm3y_fcc_ws      = dfm3y_fcc_ws.rename(columns    ={"pycno_rate": "M3Y-fcc-ws"})
dfm3y_fcc_relax   = dfm3y_fcc_relax.rename(columns ={"pycno_rate": "M3Y-fcc-relaxed"})

dfrmf_spvh        = dfrmf_spvh.rename(columns      ={"pycno_rate": "RMF-SPVH"})
dfrmf_bcc_static  = dfrmf_bcc_static.rename(columns={"pycno_rate": "RMF-bcc-static"})
dfrmf_bcc_ws      = dfrmf_bcc_ws.rename(columns    ={"pycno_rate": "RMF-bcc-ws"})
dfrmf_bcc_relax   = dfrmf_bcc_relax.rename(columns ={"pycno_rate": "RMF-bcc-relaxed"})
dfrmf_fcc_static  = dfrmf_fcc_static.rename(columns={"pycno_rate": "RMF-fcc-static"})
dfrmf_fcc_ws      = dfrmf_fcc_ws.rename(columns    ={"pycno_rate": "RMF-fcc-ws"})
dfrmf_fcc_relax   = dfrmf_fcc_relax.rename(columns ={"pycno_rate": "RMF-fcc-relaxed"})
```

```python
thisXlim = [0.8e11,1.0e11]
ax = dfm3y_spvh.plot(       x='density', y='M3Y-SPVH'      ,style='k--',xlim=thisXlim)
dfsaopaulo_spvh.plot(ax=ax, x='density', y='Sao Paulo-SPVH',style='g-.',xlim=thisXlim)
dfrmf_spvh.plot     (ax=ax, x='density', y='RMF-SPVH'      ,style='r-' ,xlim=thisXlim)

ax.set_xlabel("Density (g/cm^3)")
ax.set_ylabel("Rxn Rate (count/cm^3/sec)")
ax.set_title("SPVH - Rxn Rate vs. Density")
plt.savefig(imagesDir + '/' + 'SPVH_NN_Rates.eps')
```
