```python
import pandas as pd
import numpy as np
```

```python
dfsaopaulo = pd.read_csv("../researchdata/pycno_rates_C-C_WD_SaoPaulo.csv",index_col=False,skipinitialspace=True)
dfm3y = pd.read_csv("../researchdata/pycno_rates_C-C_WD_M3Y.csv",index_col=False,skipinitialspace=True)
dfsaopaulo['nninteraction'] = dfsaopaulo['nninteraction'].str.strip()
dfm3y['nninteraction'] = dfm3y['nninteraction'].str.strip()
```

```python
dfsaopaulo_spvh = dfsaopaulo.query('nninteraction == "SAOPAULO-SPVH"')
dfsaopaulo_bcc_static = dfsaopaulo.query('nninteraction == "SAOPAULO-bcc-static"')
dfsaopaulo_bcc_ws = dfsaopaulo.query('nninteraction == "SAOPAULO-bcc-ws"')
dfsaopaulo_bcc_relax = dfsaopaulo.query('nninteraction == "SAOPAULO-bcc-relax"')
dfsaopaulo_fcc_static = dfsaopaulo.query('nninteraction == "SAOPAULO-fcc-static"')
dfsaopaulo_fcc_ws = dfsaopaulo.query('nninteraction == "SAOPAULO-fcc-ws"')
dfsaopaulo_fcc_relax = dfsaopaulo.query('nninteraction == "SAOPAULO-fcc-relax"')

dfm3y_spvh = dfm3y.query('nninteraction == "M3Y-SPVH"')
dfm3y_bcc_static = dfm3y.query('nninteraction == "M3Y-bcc-static"')
dfm3y_bcc_ws = dfm3y.query('nninteraction == "M3Y-bcc-ws"')
dfm3y_bcc_relax = dfm3y.query('nninteraction == "M3Y-bcc-relax"')
dfm3y_fcc_static = dfm3y.query('nninteraction == "M3Y-fcc-static"')
dfm3y_fcc_ws = dfm3y.query('nninteraction == "M3Y-fcc-ws"')
dfm3y_fcc_relax = dfm3y.query('nninteraction == "M3Y-fcc-relax"')

dfsaopaulo_fcc_relax = dfsaopaulo_fcc_relax.rename(columns={"pycno_rate": "Sao Paulo-fcc-relaxed"})
dfm3y_fcc_relax = dfm3y_fcc_relax.rename(columns={"pycno_rate": "M3Y-fcc-relaxed"})
dfm3y_bcc_relax = dfm3y_bcc_relax.rename(columns={"pycno_rate": "M3Y-bcc-relaxed"})
dfm3y_spvh = dfm3y_spvh.rename(columns={"pycno_rate": "M3Y-SPVH"})
```

```python
ax = dfm3y_fcc_relax.plot(x='density', y='M3Y-fcc-relaxed',logx='true', logy='true')
dfm3y_spvh.plot(ax=ax, x='density', y='M3Y-SPVH',logx='true', logy='true')
ax.set_xlabel("Log Density (g/cm^3)")
ax.set_ylabel("Log Rxn Rate (count/cm^3/sec)")
ax.set_title("Rxn Rate vs. Density")
```

```python
ax = dfm3y_fcc_relax.plot(x='density', y='M3Y-fcc-relaxed',logx='true', logy='true')
dfsaopaulo_fcc_relax.plot(ax=ax, x='density', y='Sao Paulo-fcc-relaxed',logx='true', logy='true')
ax.set_xlabel("Log Density (g/cm^3)")
ax.set_ylabel("Log Rxn Rate (count/cm^3/sec)")
ax.set_title("Rxn Rate vs. Density")
```

```python
ax = dfm3y_fcc_relax.plot(x='density', y='M3Y-fcc-relaxed',logx='true', logy='true')
dfm3y_bcc_relax.plot(ax=ax, x='density', y='M3Y-bcc-relaxed',logx='true', logy='true')
ax.set_xlabel("Log Density (g/cm^3)")
ax.set_ylabel("Log Rxn Rate (count/cm^3/sec)")
ax.set_title("Rxn Rate vs. Density")
```

```python
dfsaopaulo.plot(x='density', y='pycno_rate', logx='true', logy='true', title="Sao Paulo Rxn Rate vs Density")
dfm3y.plot(x='density', y='pycno_rate', logx='true', logy='true', title="M3Y Rxn Rate vs Density")
```
